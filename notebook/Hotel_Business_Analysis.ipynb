```python
# Define correct calendar month order

month_order = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]
```

```python
# Calculate monthly bookings for each hotel type

monthly_bookings = (
    df.groupby(["arrival_date_month", "hotel"])
      .size()
      .reset_index(name="bookings")
)

monthly_bookings["arrival_date_month"] = pd.Categorical(
    monthly_bookings["arrival_date_month"],
    categories=month_order,
    ordered=True
)

monthly_bookings = monthly_bookings.sort_values(
    "arrival_date_month"
)

monthly_bookings
```

## 2.1 Monthly Booking Analysis by Hotel Type

### Question 3 — Seasonal Pattern

The monthly booking pattern shows clear seasonal variation in hotel
demand. Booking volumes are higher during the peak travel months
and lower during quieter months.

This pattern can be influenced by holidays, vacation periods,
weather conditions, and differences between leisure and business
travel. City Hotel may also benefit from business and city-related
travel, while Resort Hotel demand can be more influenced by leisure
and holiday travel.

### Question 4 — Business Decisions

The hotel can use the monthly booking trend to plan its operations
and revenue strategy.

During peak-demand months, the hotel can increase staffing,
optimize room availability, and use demand-based pricing.

During quieter months, the hotel can introduce special offers,
discounts, seasonal packages, and targeted marketing campaigns to
increase bookings.

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_bookings,
    x="arrival_date_month",
    y="bookings",
    hue="hotel",
    marker="o"
)

plt.title("Monthly Bookings by Hotel Type")
plt.xlabel("Month")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

**Caption:** This chart compares monthly booking volumes for City Hotel and Resort Hotel and highlights seasonal changes in demand.

```python
# Find the busiest month for each hotel type

busiest_month = (
    monthly_bookings.loc[
        monthly_bookings.groupby("hotel")["bookings"].idxmax()
    ]
)

busiest_month
```

```python
# Find the quietest month for each hotel type

quietest_month = (
    monthly_bookings.loc[
        monthly_bookings.groupby("hotel")["bookings"].idxmin()
    ]
)

quietest_month
```

```python
# Calculate cancellation rate for each hotel type

cancellation_by_hotel = (
    df.groupby("hotel")["is_canceled"]
      .mean()
      .mul(100)
      .reset_index(name="cancellation_rate")
)

cancellation_by_hotel
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=cancellation_by_hotel,
    x="hotel",
    y="cancellation_rate"
)

plt.title("Cancellation Rate by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Cancellation Rate (%)")

plt.tight_layout()
plt.show()
```

**Caption:** This chart shows how the cancellation rate changes with total stay duration.

```python
# Calculate cancellation rate for each stay duration and hotel type

stay_cancellation = (
    df.groupby(["hotel", "total_stay"])["is_canceled"]
      .mean()
      .mul(100)
      .reset_index(name="cancellation_rate")
)

stay_cancellation.head()
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=stay_cancellation,
    x="total_stay",
    y="cancellation_rate",
    hue="hotel",
    marker="o"
)

plt.title("Cancellation Rate vs Total Stay Duration")
plt.xlabel("Total Stay (Nights)")
plt.ylabel("Cancellation Rate (%)")

plt.tight_layout()
plt.show()
```

```python
# Find lowest and highest cancellation rates

lowest = stay_cancellation.loc[
    stay_cancellation["cancellation_rate"].idxmin()
]

highest = stay_cancellation.loc[
    stay_cancellation["cancellation_rate"].idxmax()
]

print("Lowest Cancellation Rate:")
print("Stay:", lowest["total_stay"], "nights")
print("Rate:", round(lowest["cancellation_rate"], 2), "%")

print("\nHighest Cancellation Rate:")
print("Stay:", highest["total_stay"], "nights")
print("Rate:", round(highest["cancellation_rate"], 2), "%")
```

```python
# Cancellation rate by total stay and hotel type

stay_hotel_cancellation = (
    df.groupby(["total_stay", "hotel"])["is_canceled"]
      .mean()
      .reset_index()
)

stay_hotel_cancellation["cancellation_rate"] = (
    stay_hotel_cancellation["is_canceled"] * 100
)

stay_hotel_cancellation = stay_hotel_cancellation[
    ["total_stay", "hotel", "cancellation_rate"]
]

stay_hotel_cancellation.head(10)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=stay_hotel_cancellation,
    x="total_stay",
    y="cancellation_rate",
    hue="hotel",
    marker="o"
)

plt.title("Cancellation Rate vs Total Stay by Hotel Type")
plt.xlabel("Total Stay Duration (Nights)")
plt.ylabel("Cancellation Rate (%)")

plt.tight_layout()
plt.show()
```

**Caption:** This chart compares cancellation rates across different stay durations for City Hotel and Resort Hotel.

```python
# Calculate correlation between stay duration and cancellation

stay_correlation = df[
    ["total_stay", "is_canceled"]
].corr()

stay_correlation
```

```python
# Calculate cancellation rate for each lead time and hotel type

lead_time_cancellation = (
    df.groupby(["hotel", "lead_time"])["is_canceled"]
      .mean()
      .mul(100)
      .reset_index(name="cancellation_rate")
)

lead_time_cancellation.head()
```

```python
plt.figure(figsize=(14, 6))

sns.lineplot(
    data=lead_time_cancellation,
    x="lead_time",
    y="cancellation_rate",
    hue="hotel"
)

plt.title("Cancellation Rate vs Lead Time")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Cancellation Rate (%)")

plt.tight_layout()
plt.show()
```

**Caption:** This chart shows how cancellation rate changes as the number of days between booking and arrival increases.

```python
# Find the lead time with the lowest cancellation rate

lowest_cancellation = (
    lead_time_cancellation.loc[
        lead_time_cancellation.groupby("hotel")["cancellation_rate"].idxmin()
    ]
)

lowest_cancellation
```

```python
# Find the lead time with the highest cancellation rate

highest_cancellation = (
    lead_time_cancellation.loc[
        lead_time_cancellation.groupby("hotel")["cancellation_rate"].idxmax()
    ]
)

highest_cancellation
```

## Stage 3 — Summary and Recommendations

## 3.1 Key Findings

### 1. Hotel Type Popularity

City Hotel is the most popular hotel type and receives more bookings
than Resort Hotel. This indicates that City Hotel has a stronger
booking demand in the analyzed dataset.

### 2. Seasonality

The booking pattern shows clear seasonal variation. August is the
busiest month, while January is the quietest month in terms of total
bookings.

This indicates that hotel demand changes throughout the year and
seasonal patterns should be considered while planning pricing,
staffing, and marketing activities.

### 3. Cancellation Behaviour

The overall cancellation rate is approximately 37.04%, meaning that
a significant proportion of bookings are cancelled.

City Hotel has a higher cancellation rate than Resort Hotel,
indicating a greater cancellation risk for City Hotel bookings.

### 4. Stay Duration

The analysis of total stay duration shows that cancellation behaviour
changes with the length of stay. Longer bookings should therefore be
monitored carefully because changes in stay duration can be associated
with changes in cancellation behaviour.

### 5. Lead Time

The analysis shows that cancellation behaviour is strongly related
to lead time. Bookings made further in advance tend to have a higher
cancellation risk, particularly for City Hotel.

Therefore, lead time is an important factor that the hotel should
consider when managing cancellation risk.

## 3.3 Recommendations Based on Stay Duration and Cancellations

The hotel should monitor cancellation behaviour for longer stays
and review its cancellation policies accordingly.

The hotel can consider:

- Clear cancellation terms for longer bookings
- Appropriate cancellation charges
- Special pricing for confirmed longer stays
- Minimum-stay policies during high-demand periods

These measures can help reduce potential revenue loss caused by
cancelled bookings.

## 3.4 Recommendations Based on Lead Time

Since bookings made far in advance can have a higher cancellation
risk, the hotel should take additional measures for such bookings.

Recommended actions include:

- Sending booking confirmation and reminder messages
- Using deposits for selected advance bookings
- Providing controlled rescheduling options
- Offering suitable pricing incentives for confirmed bookings

These actions can help reduce cancellations and protect future
revenue.

## 3.5 Most Important Recommendation

The most impactful recommendation is to reduce cancellation risk
for bookings made far in advance.

The analysis shows that lead time is an important factor associated
with cancellation behaviour. Therefore, the hotel should introduce
confirmation reminders, suitable deposit policies, and flexible
rescheduling options for advance bookings.

Reducing cancellations can directly protect room availability and
hotel revenue. Therefore, managing cancellation risk for
far-ahead bookings should be a major priority.
